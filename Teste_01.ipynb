```python
import unittest

# Função que multiplica dois números
def multiplicar(a, b):
    return a * b

# Classe de teste que verifica a função multiplicar
class TestMultiplicacao(unittest.TestCase):

    # Testa multiplicação de dois números positivos
    def test_multiplicacao_positiva(self):
        self.assertEqual(multiplicar(3, 4), 12)

    # Testa multiplicação quando um dos números é zero
    def test_multiplicacao_por_zero(self):
        self.assertEqual(multiplicar(5, 0), 0)

# Executa os testes no ambiente Colab/Jupyter
unittest.main(argv=[''], exit=False)
```
