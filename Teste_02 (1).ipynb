```python
# Função que verifica se uma string é palíndromo (ignora espaços e caixa)
def eh_palindromo(s):
    s = s.lower().replace(" ", "")  # Remove espaços e coloca em minúsculas
    return s == s[::-1]             # Compara string com sua inversa

# Teste para palavra palíndromo simples
def test_palindromo_simples():
    assert eh_palindromo("arara") == True
    print("Teste palíndromo simples passou")

# Teste para frase palíndromo com espaços
def test_palindromo_com_espaco():
    assert eh_palindromo("A mala nada na lama") == True
    print("Teste palíndromo com espaço passou")

# Teste para palavra que não é palíndromo
def test_palindromo_falso():
    assert eh_palindromo("python") == False
    print("Teste palíndromo falso passou")

# Executa os três testes
test_palindromo_simples()
test_palindromo_com_espaco()
test_palindromo_falso()
```
