```python
import requests

# Função que verifica se a página da Google responde com status HTTP 200
def test_api_google():
    response = requests.get("https://www.google.com")  # Faz a requisição GET
    assert response.status_code == 200                 # Verifica se o status é 200
    print("Teste API Google passou: status 200 confirmado")

# Executa o teste
test_api_google()
```
